Bài 6

```python
import torch

a = torch.randn(2, 3)
b = torch.randn(2, 3)

print("a:")
print(a)

print("b:")
print(b)

print("\na + b:")
print(a + b)

print("\na / b:")
print(a / b)

print("\na * b:")
print(a * b)

print("\na ** 2:")
print(a ** 2)

print("\na @ b.T:")
print(a @ b.T)

print("Shape:", (a @ b.T).shape)
```

* Phân biệt a * b và a @ b.T
- a * b: là nhân từng phần tử:
+ a: 2 × 3
+ b: 2 × 3
+ a * b: 2 × 3
- Ví dụ:
+ a = [1  2  3]
+ b = [4  5  6]
+ a * b = [4  10  18]
- a @ b.T là phép nhân ma trận: 
+ a   = 2 × 3
+ b.T = 3 × 2
+ (2 × 3) @ (3 × 2)
       ↓
     2 × 2
- Vì vậy:
+ a * b
→ torch.Size([2, 3])

+ a @ b.T
→ torch.Size([2, 2])

```python
import torch

a = torch.randn(3, 4)

print("a:")
print(a)

print("\nMean:")
print(torch.mean(a, dim=0))

print("\nStandard deviation:")
print(torch.std(a, dim=0))

print("\nCumulative sum:")
print(torch.cumsum(a, dim=0))
print("\nMean dim=1:")
print(torch.mean(a, dim=1))

print("\nStandard deviation dim=1:")
print(torch.std(a, dim=1))

print("\nCumulative sum dim=1:")
print(torch.cumsum(a, dim=1))
```

Với a = 3 × 4
* dim = 0
- Tính theo cột:
+ Mean → 4 giá trị
+ Std  → 4 giá trị
* dim = 1
- Tính theo dòng:
+ Mean → 3 giá trị
+ Std  → 3 giá trị
* dim=0 thực hiện phép tính theo chiều đầu tiên, tương ứng tính theo từng cột của Tensor. dim=1 thực hiện phép tính theo chiều thứ hai, tương ứng tính theo từng dòng. Vì vậy số lượng giá trị kết quả thay đổi từ 4 sang 3.

Bài 9

```python
import numpy as np
import torch

# Da thuc y = x^2 + 2x + 3
polynomial = np.poly1d([1, 2, 3])

N = 20

# Tao du lieu
X = np.random.randn(N, 1) * 5
Y = polynomial(X)

# Tao ma tran dac trung
XX = np.hstack([
    X * X,
    X,
    np.ones_like(X)
])

# Khoi tao he so ngau nhien
w = torch.randn(3, 1, requires_grad=True)

x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)

# Optimizer
optimizer = torch.optim.NAdam([w], lr=0.01)

print("Initial coefficients:")
print(w)

for epoch in range(1000):

    optimizer.zero_grad()

    y_pred = x @ w

    mse = torch.mean(torch.square(y - y_pred))

    mse.backward()

    optimizer.step()

print("\nLearned coefficients:")
print(w)  

print("\nHe so mong doi:")
print("[1, 2, 3]")
```

- Ba hệ số học được cần gần với:
+ [1, 2, 3]
- Vì đa thức ban đầu là:
+ y = 1*x² + 2*x + 3
- vai trò của optimizer.zero_grad(), mse.backward() và optimizer.step():
+ optimizer.zero_grad():Xóa gradient của lần lặp trước.
+ mse.backward(): PyTorch dùng autograd để tính gradient của hàm lỗi mse.
+ optimizer.step(): Optimizer cập nhật các hệ số w dựa trên gradient.


Bài 10

```python
import random
import torch

A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)

EPOCHS = 2000

optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)

for _ in range(EPOCHS):

    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2

    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4

    optimizer.zero_grad()

    sqerr.backward()

    optimizer.step()

print("A =", A)
print("B =", B)
print("C =", C)
print("D =", D)

print("\nKiem tra:")
print("A + B =", A + B)
print("C - D =", C - D)
print("A + C =", A + C)
print("B - D =", B - D)
```

- Thực ra hệ phương trình này không có nhiều nghiệm độc lập. Ta có thể biến đổi:
+ A + B = 9
+ A + C = 8
- Suy ra:
+ B = 9 - A
+ C = 8 - A
- Từ: C - D = 1
- Suy ra: D = C - 1 = 7 - A
- Thay vào phương trình: B - D = 2
- Ta có:
+ (9 - A) - (7 - A) = 2 <=> 2 = 2
- Phương trình cuối không tạo thêm điều kiện cho A.
- Vì vậy hệ có vô số nghiệm:
+ A = t
+ B = 9 - t
+ C = 8 - t
+ D = 7 - t
- Ví dụ:
+ t = 4
→ A=4, B=5, C=4, D=3
- nhưng:
+ t = 3
→ A=3, B=6, C=5, D=4 cũng thỏa cả 4 phương trình.
* Đây chính là lý do bài yêu cầu nhận xét có thể có nhiều nghiệm.
